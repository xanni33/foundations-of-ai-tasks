```python
import numpy as np
import matplotlib.pyplot as plt
import torch
```

1. wizualizacja danych (imshow)

```python
def plot_function(f, title=""):
    x = np.linspace(-5, 5, 200)
    y = np.linspace(-5, 5, 200)

    X, Y = np.meshgrid(x, y)
    Z = f(X, Y)

    plt.figure(figsize=(6,5))
    plt.imshow(Z, extent=[-5,5,-5,5], origin='lower')
    plt.colorbar()
    plt.title(title)
    plt.xlabel("x0")
    plt.ylabel("x1")
    plt.show()
```

2. prosta funkcja f(x0, x1) = x0^2 + 3*x1^2
oraz jej gradient [2x0, 6x1]

```python
def f_easy(x0, x1):
    return x0**2 + 3*x1**2

plot_function(f_easy, "easy function")
```

```python
def grad_easy(x):
    x0, x1 = x
    return np.array([2*x0, 6*x1])
```

3. jak dla tego gradientu dziala SGD oraz momentum (scatterplot)

```python
def sgd(grad, start, lr=0.1, steps=30):
    x = np.array(start, dtype=float)
    path = [x.copy()]

    for _ in range(steps):
        x = x - lr * grad(x)
        path.append(x.copy())

    return np.array(path)

def momentum(grad, start, lr=0.1, beta=0.9, steps=30):
    x = np.array(start, dtype=float)
    v = np.zeros_like(x)

    path = [x.copy()]

    for _ in range(steps):
        v = beta * v + (1 - beta) * grad(x)
        x = x - lr * v
        path.append(x.copy())

    return np.array(path)

#wizualizacja trajektorii krokow
def plot_path(f, path, title=""):
    x = np.linspace(-5, 5, 200)
    y = np.linspace(-5, 5, 200)

    X, Y = np.meshgrid(x, y)
    Z = f(X, Y)

    plt.figure(figsize=(6,5))
    plt.imshow(Z, extent=[-5,5,-5,5], origin='lower', cmap='viridis')

    path = np.array(path)
    plt.plot(path[:,0], path[:,1], 'r.-')

    plt.title(title)
    plt.xlabel("x0")
    plt.ylabel("x1")
    plt.colorbar()
    plt.show()

#test dla sgd i momentum dla funkcji prostej
start = [4, 4]

path_sgd = sgd(grad_easy, start, lr=0.1)
path_mom = momentum(grad_easy, start, lr=0.1, beta=0.9) #dla 0.0 jest to sgd, dla np 0,6 mniejszy zygzak, 0.4 bardziej wygladzona

plot_path(f_easy, path_sgd, "SGD")
plot_path(f_easy, path_mom, "Momentum")
```

4. porownanie z autogradem (torch) -> "trudna" funkcja f(x0, x1) = sinx0 * cosx1 + 0.1(x0^2 + x1^2)

```python
import torch

def f_torch(x):
    return torch.sin(x[0]) * torch.cos(x[1]) + 0.1*(x[0]**2 + x[1]**2)
```

```python
#optymalizacja z sgd i momentum
def torch_optimize(start, lr=0.1, steps=50, momentum=False, beta=0.9):
    x = torch.tensor(start, dtype=torch.float32, requires_grad=True)

    if momentum:
        optimizer = torch.optim.SGD([x], lr=lr, momentum=beta)
    else:
        optimizer = torch.optim.SGD([x], lr=lr)

    path = []

    for _ in range(steps):
        optimizer.zero_grad()
        y = f_torch(x)
        y.backward()
        optimizer.step()

        path.append(x.detach().numpy().copy())

    return np.array(path)
```

```python
#wizualizacja
def f_hard_np(x0, x1):
    return np.sin(x0)*np.cos(x1) + 0.1*(x0**2 + x1**2)

#test
path_sgd = torch_optimize([3,3], lr=0.1, momentum=False)
path_mom = torch_optimize([3,3], lr=0.1, momentum=True)

plot_path(f_hard_np, path_sgd, "SGD")
plot_path(f_hard_np, path_mom, "Momentum")
```
