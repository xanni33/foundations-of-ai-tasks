```python
#25.03. & 01.04.
import numpy as np
import matplotlib.pyplot as plt
import torch

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams.update({
    "font.family": "serif",
    'mathtext.fontset': 'dejavuserif'
})

def generate_spiral_data(num_samples: int):
    N = num_samples // 2

    theta = np.sqrt(np.random.rand(N))*4*np.pi

    r_a = 4*theta + np.pi
    data_a = np.array([np.cos(theta)*r_a, np.sin(theta)*r_a]).T
    x_a = data_a + np.random.randn(N,2) * 3
    label_a = np.array([0]*N)[:, None]


    r_b = -4*theta - np.pi
    data_b = np.array([np.cos(theta)*r_b, np.sin(theta)*r_b]).T
    x_b = data_b + np.random.randn(N,2) * 3
    label_b = np.array([1]*N)[:, None]

    x = np.concatenate([x_a, x_b], axis=0)
    x /= np.std(x, axis=0)
    y = np.concatenate([label_a, label_b], axis=0)
    return x, y.squeeze()

def plot_2d_data(x, y, edgecolors: str = None, alpha: float = 0.8):
    plt.figure()
    plt.scatter(x[:, 0], x[:, 1], c=y, cmap="magma", s=50, alpha=alpha, vmin=0.0, vmax=y.max()+1, edgecolors=edgecolors,)
    plt.axis('equal')
    plt.xlabel(r"$x_0$", fontdict={'family': 'serif', 'size': 14,  'color': 'black'})
    plt.ylabel(r"$x_1$", fontdict={'family': 'serif', 'size': 14,  'color': 'black'})
    plt.legend()
    plt.show()
    plt.close()

x, y = generate_spiral_data(1000)

#dodane:
x = torch.tensor(x, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32).unsqueeze(1)

plot_2d_data(x, y)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F

plt.style.use('seaborn-v0_8-whitegrid')

# generowanie danych
x_np, y_np = generate_spiral_data(1000)

x = torch.tensor(x_np, dtype=torch.float32)
y = torch.tensor(y_np, dtype=torch.float32)

def build_backbone(num_blocks):
    layers = []

    # pierwszy blok: 2 -> 20
    layers.append(nn.Linear(2, 20))
    layers.append(nn.BatchNorm1d(20))
    layers.append(nn.GELU())

    # bloki srodkowe: 20 -> 20
    for _ in range(num_blocks - 2):
        layers.append(nn.Linear(20, 20))
        layers.append(nn.BatchNorm1d(20))
        layers.append(nn.GELU())

    # ostatni blok: 20 -> 2
    if num_blocks > 1:
        layers.append(nn.Linear(20, 2))
        layers.append(nn.BatchNorm1d(2))
        layers.append(nn.GELU())
    else:
        # jesli num_blocks = 1 to od razu 2 -> 2
        layers = []
        layers.append(nn.Linear(2, 2))
        layers.append(nn.BatchNorm1d(2))
        layers.append(nn.GELU())

    return layers

linear_h = nn.Linear(2, 1)

def head(x):
    x = linear_h(x)
    x = torch.sigmoid(x)
    return x

def train_model(num_blocks, x, y, epochs=2000, lr=1e-2):
    # zbudowanie backbone
    layers = build_backbone(num_blocks)
    # zmiana listy warstw na sekwencje
    backbone = nn.Sequential(*layers)

    # optymalizacja backbone + head
    params = list(backbone.parameters()) + list(linear_h.parameters())
    optimizer = torch.optim.Adam(params, lr=lr)

    loss_fn = nn.BCELoss()

    for epoch in range(epochs):
        # forward
        z = backbone(x)
        y_pred = head(z).squeeze()

        loss = loss_fn(y_pred, y)

        # backward
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        if (epoch+1) % 500 == 0:
            preds = (y_pred >= 0.5).float()
            acc = (preds == y).float().mean().item()
            print(f"[bloki={num_blocks}] epoka {epoch+1}, loss={loss.item():.4f}, acc={acc:.4f}")

    # koncowa accuracy
    preds = (y_pred >= 0.5).float()
    acc = (preds == y).float().mean().item()

    return backbone, acc

accuracies = {}

for blocks in [1, 2, 3]:
    print(f"\n=== Trening dla {blocks} bloków")
    backbone, acc = train_model(blocks, x, y)
    accuracies[blocks] = acc

print("\nWyniki accuracy:")
for k, v in accuracies.items():
    print(f"{k} bloków -> accuracy = {v:.4f}")

with torch.no_grad():
    transformed = backbone(x).numpy()

plot_2d_data(transformed, y_np, title="Dane po przejściu przez backbone")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

# dane
x_np, y_np = generate_spiral_data(1000)
x = torch.tensor(x_np, dtype=torch.float32)
y = torch.tensor(y_np, dtype=torch.float32)

# budowa backbone (3 bloki)
def build_backbone(num_blocks):
    layers = []

    # pierwszy blok: 2 -> 20
    layers.append(nn.Linear(2, 20))
    layers.append(nn.BatchNorm1d(20))
    layers.append(nn.GELU())

    # bloki środkowe: 20 -> 20
    for _ in range(num_blocks - 2):
        layers.append(nn.Linear(20, 20))
        layers.append(nn.BatchNorm1d(20))
        layers.append(nn.GELU())

    # ostatni blok: 20 -> 2
    if num_blocks > 1:
        layers.append(nn.Linear(20, 2))
        layers.append(nn.BatchNorm1d(2))
        layers.append(nn.GELU())
    else:
        layers = []
        layers.append(nn.Linear(2, 2))
        layers.append(nn.BatchNorm1d(2))
        layers.append(nn.GELU())

    return nn.Sequential(*layers)

backbone = build_backbone(3)

# head
linear_h = nn.Linear(2, 1)

def head(x):
    return torch.sigmoid(linear_h(x))

# trening
params = list(backbone.parameters()) + list(linear_h.parameters())
optimizer = torch.optim.Adam(params, lr=1e-2)
loss_fn = nn.BCELoss()

for epoch in range(2000):
    z = backbone(x)
    y_pred = head(z).squeeze()

    loss = loss_fn(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# wizualizacja danych po backbone
with torch.no_grad():
    z = backbone(x).numpy()

plt.figure(figsize=(7,7))
plt.scatter(z[:,0], z[:,1], c=y_np, cmap="magma", s=40)
plt.title("Dane po przejściu przez backbone (3 bloki)")
plt.xlabel("z0")
plt.ylabel("z1")

# granica decyzji head
W = linear_h.weight.detach().numpy().flatten()
b = linear_h.bias.detach().numpy().item()

# prosta: w0*x0 + w1*x1 + b = 0  ->  x1 = -(w0*x0 + b)/w1
x0_line = np.linspace(z[:,0].min(), z[:,0].max(), 200)
x1_line = -(W[0] * x0_line + b) / W[1]

plt.plot(x0_line, x1_line, color="cyan", linewidth=2, label="granica decyzji head")
plt.legend()
plt.show()
```

```python
# wer. bez polaczen rezydualnych
# 10-krotnie linear -> batchnorm -> gelu
def build_deep_backbone_no_residual(depth):
    layers = []

    # pierwszy blok: 2 -> 20
    layers.append(nn.Linear(2, 20))
    layers.append(nn.BatchNorm1d(20))
    layers.append(nn.GELU())

    # kolejne bloki: 20 -> 20
    for _ in range(depth - 1):
        layers.append(nn.Linear(20, 20))
        layers.append(nn.BatchNorm1d(20))
        layers.append(nn.GELU())

    # ostatnia warstwa: 20 -> 2
    layers.append(nn.Linear(20, 2))
    layers.append(nn.BatchNorm1d(2))
    layers.append(nn.GELU())

    return nn.Sequential(*layers)
```

```python
#2. wer. z polaczeniami rezydualnymi
# linear -> batchnorm -> gelu + skip connection (( block(x) -> x + F(x) ))
class ResidualBlock(nn.Module):
    def __init__(self, dim=20):
        super().__init__()
        self.linear = nn.Linear(dim, dim)
        self.bn = nn.BatchNorm1d(dim)

    def forward(self, x):
        return x + F.gelu(self.bn(self.linear(x)))


def build_deep_backbone_residual(depth):
    layers = []

    # pierwszy blok: 2 -> 20
    layers.append(nn.Linear(2, 20))
    layers.append(nn.BatchNorm1d(20))
    layers.append(nn.GELU())

    # residual blocks
    for _ in range(depth - 1):
        layers.append(ResidualBlock(20))

    # ostatnia warstwa: 20 -> 2
    layers.append(nn.Linear(20, 2))
    layers.append(nn.BatchNorm1d(2))
    layers.append(nn.GELU())

    return nn.Sequential(*layers)
```

```python
#head, trening i wizualizacja pozostają takie same, wystarczy podmienić backbone na odpowiedni
linear_h = nn.Linear(2, 1)

def head(x):
    return torch.sigmoid(linear_h(x))

def train_model(backbone, x, y, epochs=2000, lr=1e-2):
    params = list(backbone.parameters()) + list(linear_h.parameters())
    optimizer = torch.optim.Adam(params, lr=lr)
    loss_fn = nn.BCELoss()

    for epoch in range(epochs):
        z = backbone(x)
        y_pred = head(z).squeeze()

        loss = loss_fn(y_pred, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    return backbone

def plot_backbone_and_decision(backbone, x, y):
    with torch.no_grad():
        z = backbone(x).numpy()

    plt.figure(figsize=(7,7))
    plt.scatter(z[:,0], z[:,1], c=y.numpy(), cmap="magma", s=40)

    # granica decyzji head
    W = linear_h.weight.detach().numpy().flatten()
    b = linear_h.bias.detach().numpy().item()

    x0_line = np.linspace(z[:,0].min(), z[:,0].max(), 200)
    x1_line = -(W[0] * x0_line + b) / W[1]

    plt.plot(x0_line, x1_line, color="cyan", linewidth=2)
    plt.title("Backbone + granica decyzji head")
    plt.show()
```

```python
backbone_no_res = build_deep_backbone_no_residual(depth=10)
backbone_no_res = train_model(backbone_no_res, x, y)
plot_backbone_and_decision(backbone_no_res, x, y)
```

```python
backbone_res = build_deep_backbone_residual(depth=10)
backbone_res = train_model(backbone_res, x, y)
plot_backbone_and_decision(backbone_res, x, y)
```
